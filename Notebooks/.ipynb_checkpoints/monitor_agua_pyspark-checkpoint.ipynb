## 1. Crear la SparkSessionPunto de entrada a Spark. Corre en modo local usando todos los núcleos disponibles.

```python
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("monitor-agua-horeb")
    .master("local[*]")
    .config("spark.ui.port", "4040")
    .getOrCreate()
)

spark
```

```python
from pyspark.sql.types import (StructType, StructField, StringType,
    DoubleType, IntegerType)

# Definimos el esquema columna por columna
esquema = StructType([
    StructField("device_id",         StringType(),  True),
    StructField("timestamp",         StringType(),  True),
    StructField("distancia_cm",      DoubleType(),  True),
    StructField("nivel_porcentaje",  DoubleType(),  True),
    StructField("litros_actuales",   IntegerType(), True),
    StructField("capacidad_litros",  IntegerType(), True),
    StructField("bateria",           IntegerType(), True),
    StructField("zona",              StringType(),  True),
    StructField("ip",                StringType(),  True),
])

df = spark.read.csv(
    "../data/lecturas_tanques.csv",
    header=True,
    schema=esquema,
)

print("Filas:", df.count(), "| Columnas:", len(df.columns))
```

## 3. Transformaciones y evaluación perezosa (equivale a 3.5)Aplicamos `select()` y `filter()`. **Las transformaciones no ejecutan nada** — solo construyen el plan. La ejecución recién ocurre con una acción.

```python
from pyspark.sql.functions import col# Transformación: nos quedamos con lecturas de nivel crítico (< 30%)df_critico = (    df    .select("device_id", "timestamp", "nivel_porcentaje", "zona")    .filter(col("nivel_porcentaje") < 30))# Hasta aquí Spark SOLO construyó el plan. No hay salida, no leyó datos.df_critico
```

```python
# Acción: AQUÍ recién Spark ejecuta todo el plandf_critico.show(truncate=False)print("Lecturas en nivel crítico:", df_critico.count())
```

**Evaluación perezosa comprobada:** la celda anterior con solo la transformación no mostró ningún dato. Esta celda, con la acción `show()`/`count()`, sí ejecutó y devolvió resultados. Esa es la diferencia entre transformación y acción.

## 4. Analizar el plan de ejecución con explain() (equivale a 3.6)`explain(True)` muestra las 4 fases del optimizador Catalyst. Buscamos que el `Filter` haya sido movido para leer menos datos (*predicate pushdown*).

```python
df_critico.explain(True)
```

**Optimización identificada:** en el *Optimized Logical Plan*, el `Filter` (nivel < 30) aparece empujado hacia la lectura del archivo. Catalyst reorganiza el plan para filtrar lo antes posible y procesar menos datos. Esto es *predicate pushdown* en acción.

## 5. Funciones y creación de columnas con withColumn() (equivale a 3.7)Creamos columnas nuevas aplicando funciones a los datos del sensor.

```python
from pyspark.sql.functions import when, lit, to_timestamp, hour# 1. Convertir el timestamp de texto a tipo fecha realdf = df.withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss"))# 2. Clasificar el estado del tanque con when()/otherwise()df = df.withColumn(    "estado",    when(col("nivel_porcentaje") < 15, "CRITICO")    .when(col("nivel_porcentaje") < 30, "BAJO")    .when(col("nivel_porcentaje") < 60, "MEDIO")    .otherwise("LLENO"))# 3. Extraer la hora del día (útil para detectar fugas nocturnas)df = df.withColumn("hora", hour(col("timestamp")))# 4. Marcar si es horario de madrugada (cuando NO debería haber consumo)df = df.withColumn(    "es_madrugada",    when((col("hora") >= 0) & (col("hora") <= 5), lit(True)).otherwise(lit(False)))# 5. Columna constante con el nombre del proyectodf = df.withColumn("proyecto", lit("Monitor Agua Horeb"))df.select("device_id", "timestamp", "nivel_porcentaje", "estado",          "hora", "es_madrugada").show(10, truncate=False)
```

## 6. Tratamiento de calidad de datos (nulos)Los datos reales tienen lecturas perdidas. Identificamos y tratamos los nulos.

```python
# Contar nulos por columnafrom pyspark.sql.functions import count as spark_count, when as spark_whendf.select([    spark_count(spark_when(col(c).isNull(), c)).alias(c)    for c in ["nivel_porcentaje", "litros_actuales", "bateria"]]).show()
```

```python
# Filtrar las filas con nivel nulo (lecturas perdidas del sensor)print("Filas con nivel_porcentaje nulo:")df.filter(col("nivel_porcentaje").isNull()).select(    "device_id", "timestamp", "nivel_porcentaje", "bateria").show(truncate=False)# Para el análisis, nos quedamos solo con lecturas válidasdf_limpio = df.filter(col("nivel_porcentaje").isNotNull())print("Filas válidas tras limpiar:", df_limpio.count())
```

## 7. Agrupaciones y agregaciones (equivale a 3.9)Resumen agregado relevante a nuestro caso: estadísticas de consumo por tanque y por zona.

```python
from pyspark.sql.functions import avg, min as spark_min, max as spark_max, count# Resumen por tanque: nivel promedio, mínimo, y número de lecturasresumen_tanque = df_limpio.groupBy("device_id").agg(    avg("nivel_porcentaje").alias("nivel_promedio"),    spark_min("nivel_porcentaje").alias("nivel_minimo"),    spark_max("nivel_porcentaje").alias("nivel_maximo"),    count("*").alias("num_lecturas"))resumen_tanque.show(truncate=False)
```

```python
# Agrupar por ZONA: consumo promedio y batería promedio por sector de la ciudadresumen_zona = df_limpio.groupBy("zona").agg(    avg("nivel_porcentaje").alias("nivel_promedio_zona"),    avg("bateria").alias("bateria_promedio"),    count("*").alias("lecturas"))resumen_zona.show(truncate=False)
```

```python
# Agrupar por estado del tanque: cuántas lecturas hay en cada nivel de riesgodf_limpio.groupBy("estado").agg(    count("*").alias("cantidad")).orderBy("cantidad", ascending=False).show()
```

## 8. Caso de aplicación: detección de fugaNuestro diferenciador. Buscamos consumo anómalo en horario de madrugada, donde no debería haber uso de agua.

```python
# Lecturas de madrugada con nivel más bajo de lo esperado = posible fugaprint("Lecturas en horario de madrugada (posibles fugas):")df_limpio.filter(col("es_madrugada") == True).select(    "device_id", "timestamp", "nivel_porcentaje", "zona", "estado").orderBy("nivel_porcentaje").show(truncate=False)
```

**Hallazgo:** el `tanque_03` aparece en la madrugada con un nivel muy bajo (~15%), habiendo caído bruscamente respecto a sus lecturas previas de la noche. Como a esas horas no debería haber consumo, el sistema lo marca como **posible fuga** — exactamente el tipo de anomalía que el proyecto busca detectar en tiempo real.
